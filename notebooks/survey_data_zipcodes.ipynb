```python
df = pd.read_csv("../Datasets/final_data/2024_AR_longitudinal_final.csv")
df
```

```python
affected_zips = [int(x) for x in df["inputzip"].unique() if not math.isnan(x)]
affected_zips.sort()
affected_zips = [str(x) for x in affected_zips]
len(affected_zips)
```

```python
path_zips = "/glade/work/cbecker/california_zip.geojson"
gdf_zips = gpd.read_file(path_zips)
gdf_zips.tail()
```

```python
m = folium.Map(location=[34.0549, -118.2426],
               zoom_start=8)

folium.GeoJson(
    gdf_zips[gdf_zips["ZIP_CODE"].isin(affected_zips)],
    style_function=lambda feature: {
        "fillColor": "blue",
        "fillOpacity": 0.7,
        "color": "black",
        "weight": 0.5,
    },
).add_to(m)

m.save(f"./html/survey_zipcodes.html")
m
```
