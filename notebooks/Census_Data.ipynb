## CENSUS data download with Geojsons 

```python
import requests
import geopandas as gpd
import pandas as pd
import os
import zipfile
import certifi

# ========== CONFIG ==========
API_KEY = '52bed1f41c2b258689638f0261f33793016d07c2'  # 
STATE_FIPS = '06'  # California

COUNTY_CODES = {
    'Los Angeles': '037',
    'Ventura': '111',
    'Santa Barbara': '083'
}

ACS_YEAR = '2020'

DEM_VARS = {
    'B01003_001E': 'total_population',
    'B19013_001E': 'median_income',
    'B02001_002E': 'white_alone',
    'B02001_003E': 'black_alone',
    'B03003_003E': 'hispanic_latino'
}
# ============================


def fetch_demographics():
    """Fetch ACS demographic data for each county."""
    base_url = f'https://api.census.gov/data/{ACS_YEAR}/acs/acs5'
    all_dfs = []

    for county_name, county_fips in COUNTY_CODES.items():
        params = {
            'get': ','.join(DEM_VARS.keys()),
            'for': 'tract:*',
            'in': f'state:{STATE_FIPS} county:{county_fips}',
            'key': API_KEY
        }
        print(f"📊 Fetching demographic data for {county_name}...")
        response = requests.get(base_url, params=params)
        response.raise_for_status()
        data = response.json()
        df = pd.DataFrame(data[1:], columns=data[0])
        df.rename(columns=DEM_VARS, inplace=True)
        df['county_name'] = county_name
        all_dfs.append(df)

    return pd.concat(all_dfs, ignore_index=True)


def fetch_tract_geometries():
    """Download and extract entire CA shapefile and filter counties."""
    url = f"https://www2.census.gov/geo/tiger/TIGER2020/TRACT/tl_2020_{STATE_FIPS}_tract.zip"
    base_dir = "shapefiles"
    os.makedirs(base_dir, exist_ok=True)
    local_zip = os.path.join(base_dir, "tl_2020_CA_tract.zip")
    extract_dir = os.path.join(base_dir, "tl_2020_CA_tract")

    # Download if needed
    if not os.path.exists(local_zip):
        print(f"📥 Downloading full California tract shapefile...")
        r = requests.get(url, verify=certifi.where())
        if r.status_code == 200:
            with open(local_zip, "wb") as f:
                f.write(r.content)
        else:
            raise Exception(f"❌ Failed to download shapefile: {r.status_code}")

    if not os.path.exists(extract_dir):
        with zipfile.ZipFile(local_zip, 'r') as zip_ref:
            zip_ref.extractall(extract_dir)

    shp_path = [f for f in os.listdir(extract_dir) if f.endswith(".shp")][0]
    full_gdf = gpd.read_file(os.path.join(extract_dir, shp_path))

    # Filter by relevant counties
    valid_fips = list(COUNTY_CODES.values())
    filtered_gdf = full_gdf[full_gdf['COUNTYFP'].isin(valid_fips)].copy()

    # Add county name for merge
    fips_to_name = {v: k for k, v in COUNTY_CODES.items()}
    filtered_gdf['county_name'] = filtered_gdf['COUNTYFP'].map(fips_to_name)

    return filtered_gdf


def merge_geo_and_demo():
    demo_df = fetch_demographics()
    geo_gdf = fetch_tract_geometries()

    demo_df['GEOID'] = demo_df['state'] + demo_df['county'] + demo_df['tract']
    demo_df.drop(columns=['state', 'county', 'tract'], inplace=True)

    merged = geo_gdf.merge(demo_df, on="GEOID", how="left")

    # If 'county_name' got dropped, remap using COUNTYFP
    if 'county_name' not in merged.columns:
        fips_to_name = {v: k for k, v in COUNTY_CODES.items()}
        merged['county_name'] = merged['COUNTYFP'].map(fips_to_name)

    return merged



# === MAIN EXECUTION ===
if __name__ == "__main__":
    gdf = merge_geo_and_demo()

    print("✅ Sample output:")
    print(gdf[['GEOID', 'county_name', 'total_population', 'median_income']].head())

    # Save outputs
    gdf.to_file("tracts_demographics.geojson", driver='GeoJSON')
    gdf.drop(columns='geometry').to_csv("tracts_demographics.csv", index=False)
```

## Data download code test

```python
import requests
import geopandas as gpd
import pandas as pd
import os
import zipfile
import certifi
from urllib.request import urlretrieve

# ========== CONFIG ==========
API_KEY = '52bed1f41c2b258689638f0261f33793016d07c2'
STATE_FIPS = '06'  # California
COUNTIES = {
    'Los Angeles': '037',
    'Ventura': '111',
    'Santa Barbara': '083'
}
ACS_YEAR = '2020'
DEM_VARS = {
    'B01003_001E': 'total_population',
    'B19013_001E': 'median_income',
    'B02001_002E': 'white_alone',
    'B02001_003E': 'black_alone',
    'B03003_003E': 'hispanic_latino'
}
# ============================

def fetch_demographics():
    """Fetch ACS demographic data for each county"""
    print("Fetching ACS demographic data...")
    base_url = f'https://api.census.gov/data/{ACS_YEAR}/acs/acs5'
    all_dfs = []
    for county_name, county_fips in COUNTIES.items():
        print(f" Fetching demographic data for {county_name}...")
        params = {
            'get': ','.join(DEM_VARS.keys()),
            'for': 'tract:*',
            'in': f'state:{STATE_FIPS} county:{county_fips}',
            'key': API_KEY
        }
        response = requests.get(base_url, params=params)
        response.raise_for_status()
        data = response.json()
        df = pd.DataFrame(data[1:], columns=data[0])
        df.rename(columns=DEM_VARS, inplace=True)
        df['county_name'] = county_name
        all_dfs.append(df)
    df = pd.concat(all_dfs, ignore_index=True)

    # Add GEOID
    df['GEOID'] = df['state'] + df['county'] + df['tract']

    # Convert to numeric
    for col in DEM_VARS.values():
        df[col] = pd.to_numeric(df[col], errors='coerce')

    # Add percentages
    df['pct_white'] = 100 * df['white_alone'] / df['total_population']
    df['pct_black'] = 100 * df['black_alone'] / df['total_population']
    df['pct_hispanic'] = 100 * df['hispanic_latino'] / df['total_population']

    return df


def fetch_tract_geometries():
    """Download and filter tract shapefile for selected counties in California."""
    print("Downloading full California tract shapefile...")
    ca_url = 'https://www2.census.gov/geo/tiger/TIGER2020/TRACT/tl_2020_06_tract.zip'
    local_zip = 'shapefiles/tl_2020_06_tract.zip'
    extract_dir = 'shapefiles/tl_2020_06_tract'

    os.makedirs('shapefiles', exist_ok=True)

    if not os.path.exists(local_zip):
        r = requests.get(ca_url, verify=certifi.where())
        if r.status_code == 200:
            with open(local_zip, 'wb') as f:
                f.write(r.content)
        else:
            raise RuntimeError(f"Failed to download shapefile: {r.status_code}")

    if not os.path.exists(extract_dir):
        with zipfile.ZipFile(local_zip, 'r') as zip_ref:
            zip_ref.extractall(extract_dir)

    # Load all tracts, filter to selected counties
    shp_file = [f for f in os.listdir(extract_dir) if f.endswith(".shp")][0]
    gdf = gpd.read_file(os.path.join(extract_dir, shp_file))
    selected_fips = list(COUNTIES.values())
    gdf = gdf[gdf['COUNTYFP'].isin(selected_fips)]
    gdf['county_name'] = gdf['COUNTYFP'].map({v: k for k, v in COUNTIES.items()})
    return gdf


def merge_geo_and_demo():
    """Merge geographic and demographic data into a single GeoDataFrame."""
    demo_df = fetch_demographics()
    geo_gdf = fetch_tract_geometries()

    # Drop county_name from geo_gdf to avoid column conflict
    geo_gdf = geo_gdf.drop(columns=['county_name'], errors='ignore')

    merged = geo_gdf.merge(demo_df, on="GEOID", how="left")
    return merged



# === MAIN EXECUTION ===
if __name__ == "__main__":
    gdf = merge_geo_and_demo()

    print("Sample output:")
    print(gdf[['GEOID', 'county_name', 'total_population', 'median_income', 'pct_white', 'pct_black', 'pct_hispanic']].head())

    gdf.to_file("tracts_demographics.geojson", driver='GeoJSON')
    gdf.drop(columns='geometry').to_csv("tracts_demographics.csv", index=False)
```
