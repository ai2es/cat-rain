```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_precip_LA = pd.read_parquet(f"bootstrapping_daily/results_precip_LA.parquet").fillna(value=np.nan)
df_streamflow_LA = pd.read_parquet(f"bootstrapping_daily/results_streamflow_LA.parquet").fillna(value=np.nan)
df_ffw_LA = pd.read_parquet(f"bootstrapping_daily/results_ffw_LA.parquet").fillna(value=np.nan)
df_precip_LA
```

```python
cols = [c for c in list(df_precip_LA.columns) if "_pvalue" in c]
idxs = [i for i in list(df_precip_LA.index) if "num_stops" in i]

df_precip_LA.loc[idxs, cols]
```

```python
sns.set_style(style='white')
plt.figure(figsize=(10, 9))
g1 = sns.heatmap(df_precip_LA.loc[idxs, cols],
                 cmap="coolwarm",
                 cbar_kws={'label': 'P-value', 'shrink': 0.94, 'aspect': 20, 'pad': 0.01},
                 vmin=0.0,
                 vmax=0.05,
                )
plt.show()
```
